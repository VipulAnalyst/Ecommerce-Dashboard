```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
rev = df[df['Total_Amount'] < 99999]

rev.groupby('Category')['Total_Amount'].sum().sort_values(ascending=False).plot(kind='bar', title='Revenue by Category')
plt.show()

rev.groupby('City')['Total_Amount'].sum().sort_values(ascending=False).plot(kind='bar', title='Revenue by City')
plt.show()

pd.crosstab(df['Payment_Method'], df['Order_Status'], normalize='index').plot(kind='bar', stacked=True, title='Order Status by Payment Method')
plt.show()
```

```python
aug = rev[rev['Order_Date'].dt.month == 8]
print(len(aug), "orders in August,", round(aug['Total_Amount'].sum()), "total")
print(aug.sort_values('Total_Amount', ascending=False)[['Order_ID', 'Product_Name', 'Quantity', 'Total_Amount', 'Order_Status']].head(5))
```

```python
delivered = df[df['Order_Status'] == 'Delivered']
delivered.groupby('Category')['Total_Amount'].sum().sort_values(ascending=False).plot(kind='bar', title='Delivered Revenue by Category')
plt.show()
```

```python
monthly = rev.set_index('Order_Date').resample('ME')['Total_Amount'].sum()
monthly.plot(marker='o', title='Monthly Revenue')
plt.show()
```
